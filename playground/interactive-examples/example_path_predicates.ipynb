# Path predicates and auth-domination detector

`teal_path_predicates.PathPredicateAnalysis` runs a forward dataflow
with intersection at BB joins and accumulates the branch / assert
outcomes that hold on **every** path to each BB. Edge predicates per
op:

- `bnz l` / `bz l` — `(value, "nonzero" | "zero")` on the matching side.
- `assert` — `(value, "nonzero")` on the only successor.
- `switch t0 t1 … tN-1` — target k carries `(key == k)`; fall-through
  carries `(key not in [0..N-1])`.
- `match t0 t1 … tN-1` — target k carries `(key == vk)` where `vk` is the
  k-th candidate popped from the stack; fall-through carries
  `(key not in {v0, …, vN-1})`.

`teal_auth_domination.AuthDominationDetector` consumes the analysis
to flag sensitive sinks (state-mutating ops by default) that aren't
dominated by any matcher-recognised guard. Default matcher: `txn
Sender == <bytes-const>`.

## Auth-domination detector

Pluggable: `AuthSink` picks out which assignments need guards;
`AuthMatcher` recognises a guard pattern. Built-ins: state-mutating
ops as sinks; `txn Sender == <bytes-const>` as the matcher.

```python
for case in ("vuln", "safe"):
    print(f"### {case}")
    p = teal_ssa.SSAProgram(FIXTURES / "auth_domination" / case)
    p.propagate_constants()
    vs = AuthDominationDetector(p).detect()
    print(f"  {len(vs)} violation(s)")
    for v in vs:
        print(f"    {v.pretty()}")
    print()
```

## Adding a custom matcher: `Global.CreatorAddress == Txn.Sender`

The single most common admin pattern in TEAL: only the contract's
deployer can perform sensitive actions. The default matcher recognises
``txn Sender == <bytes-const>`` (a hardcoded admin address), but most
real contracts compare against ``global CreatorAddress`` so the
authoritative address tracks the actual deployer rather than being
baked into the bytecode.

Adding the matcher is appending one entry to the matcher list — no
detector internals need to change.

```python
from tealql.tealtools.cfg.path_predicates import BranchCondition
from tealql.tealtools.ssa import SSAVar


def _is_global_creator(op) -> bool:
    if not isinstance(op, SSAVar) or op.defined_by is None:
        return False
    src = op.defined_by
    return src.op == "global" and src.immediates.strip() == "CreatorAddress"


def _is_txn_sender(op) -> bool:
    if not isinstance(op, SSAVar) or op.defined_by is None:
        return False
    src = op.defined_by
    return src.op == "txn" and src.immediates.strip() == "Sender"


def _matches_creator_check(cond: BranchCondition, prog) -> bool:
    """Recognise `Global.CreatorAddress == Txn.Sender` (either order)."""
    if cond.kind != "nonzero":
        return False
    v = cond.value
    if not isinstance(v, SSAVar) or v.defined_by is None:
        return False
    a = v.defined_by
    if a.op != "==" or len(a.inputs) != 2:
        return False
    a0, a1 = a.inputs
    return (
        (_is_global_creator(a0) and _is_txn_sender(a1))
        or (_is_global_creator(a1) and _is_txn_sender(a0))
    )


CREATOR_AUTH_MATCHER = AuthMatcher(
    name="Global.CreatorAddress == Txn.Sender",
    matches=_matches_creator_check,
)

# Re-run on the safe fixture with the combined matcher set — the
# default matcher still fires (its sender check is hardcoded), so
# this should still report 0 violations.
p = teal_ssa.SSAProgram(FIXTURES / "auth_domination" / "safe")
p.propagate_constants()
det = AuthDominationDetector(
    p, matchers=[*DEFAULT_MATCHERS, CREATOR_AUTH_MATCHER],
)
vs = det.detect()
print(f"safe with combined matchers: {len(vs)} violation(s) (expected 0)")
```
