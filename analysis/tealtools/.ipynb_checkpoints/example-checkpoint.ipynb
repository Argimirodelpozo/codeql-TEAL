# TEAL Graph Exploration

Build a NetworkX `MultiDiGraph` from a TEAL CodeQL database, walk it with standard NetworkX ops, and render slices with Graphviz `dot`.

**Dependencies:** `networkx`, `codeql` on PATH, `dot` (Graphviz) on PATH.

The first cell does a cold load (~20-30s — compiles three queries). Subsequent loads hit a per-db CSV cache.

```python
import sys
from pathlib import Path

# Make the teal_graphs module importable. Adjust if running from elsewhere.
HERE = Path.cwd()
sys.path.insert(0, str(HERE))

from teal_graphs import (
    load_graph,
    cfg_view, dataflow_view,
    draw_cfg, draw_dataflow, draw,
    to_dot,
)
import networkx as nx
```

## 1. Load a database

Point at any built TEAL CodeQL db. The repo ships two: `test-dbs/xgov-db` and `test-dbs/folks-finance-db`.

```python
DB = HERE.parent / "test-dbs" / "xgov-db"
g = load_graph(DB)
g.number_of_nodes(), g.number_of_edges()
```

## 2. Structure of a node

Nodes are keyed by `(file, line)`. Attributes: `qlClass` (the CodeQL primary class — `BnzOpcode`, `AppGlobalPutOpcode`, ...), `text` (the opcode's `toString()`).

```python
from itertools import islice

for n, data in islice(g.nodes(data=True), 5):
    print(n, data)
```

## 3. Edge kinds

Every edge has `kind` = `"cfg"` or `"dataflow"`. CFG edges additionally carry `successor` (e.g. `NormalSuccessor`, `BooleanSuccessor(true)`, `RetsubCompletion`).

```python
from collections import Counter

kind_counts = Counter(d["kind"] for _, _, d in g.edges(data=True))
successor_counts = Counter(
    d["successor"] for _, _, d in g.edges(data=True) if d["kind"] == "cfg"
)
print("edges by kind:     ", dict(kind_counts))
print("cfg by successor:  ", dict(successor_counts))
```

## 4. Kind-specific views

`cfg_view(g)` and `dataflow_view(g)` return read-only subgraphs filtered by edge kind. Any NetworkX algorithm (traversal, paths, components) works on them.

```python
cfg = cfg_view(g)
df  = dataflow_view(g)

print("cfg:     ", cfg.number_of_nodes(), "nodes,", cfg.number_of_edges(), "edges")
print("dataflow:", df.number_of_nodes(),  "nodes,", df.number_of_edges(),  "edges")
```

## 5. Walking the graph

Example: find every CFG successor of each conditional branch opcode.

```python
branches = [
    n for n, d in g.nodes(data=True)
    if d["qlClass"] in {"BnzOpcode", "BzOpcode"}
]
print(f"{len(branches)} branch opcodes\n")

for b in branches[:5]:
    succs = [(v, cfg.get_edge_data(b, v)[0]["successor"]) for v in cfg.successors(b)]
    print(f"{b} ({g.nodes[b]['qlClass']}) →")
    for v, s in succs:
        print(f"    {v}  [{s}]")
```

## 6. Shortest CFG path between two opcodes

```python
# First opcode in the program and a later one, as a demo.
entry = min(cfg.nodes, key=lambda n: (n[0], n[1]))
later = sorted(cfg.nodes, key=lambda n: (n[0], n[1]))[50]

try:
    path = nx.shortest_path(cfg, entry, later)
    print(f"{entry} → {later}: {len(path)} hops")
    for step in path:
        print(" ", step, g.nodes[step]['qlClass'])
except nx.NetworkXNoPath:
    print("no path")
```

## 7. Rendering a CFG slice with Graphviz

For 1000+ node graphs the full render is huge. Start by extracting a local neighborhood, then render that subgraph.

Below: CFG nodes within 6 CFG hops of the first opcode.

```python
start = entry
near = nx.single_source_shortest_path_length(cfg, start, cutoff=6).keys()
slice_g = g.subgraph(near).copy()

draw_cfg(slice_g)
```

## 8. Dataflow slice

Find everything reachable *from* a given opcode by local dataflow.

```python
# Pick a txn-like source to seed the slice.
sources = [n for n, d in g.nodes(data=True) if d['qlClass'] == 'TxnOpcode']
if sources:
    seed = sources[0]
    reach = nx.descendants(df, seed) | {seed}
    print(f"{seed} reaches {len(reach)} nodes via dataflow")
    draw_dataflow(g.subgraph(reach).copy())
```

## 9. Overlay: CFG + dataflow on the same slice

Solid edges = CFG, dotted blue = dataflow.

```python
draw(slice_g, kinds=("cfg", "dataflow"))
```

## 10. Full-file CFG (for small programs)

For the ~1200-node xgov contract this is huge but legible; for larger contracts prefer slicing first.

```python
files = sorted({n[0] for n in g.nodes})
print("files:", files)

# Save to disk instead of embedding a huge SVG in the notebook.
svg = draw_cfg(g, file=files[0])
out_path = Path.home() / f"teal_cfg_{Path(files[0]).stem}.svg"
svg.save(out_path)
print("saved to", out_path)
```

## 11. Lower-level: raw DOT source

If you want to post-process the DOT (e.g. group nodes by subroutine into clusters), use `to_dot` directly.

```python
dot_src = to_dot(slice_g, kinds=("cfg",))
print(dot_src[:500], "...\n")
print("total:", len(dot_src), "bytes")
```
